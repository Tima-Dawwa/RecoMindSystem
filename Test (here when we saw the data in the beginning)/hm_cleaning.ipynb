```python
import pandas as pd

# Load the original CSV file
file_path = 'data/filtered_data.csv'  # Update with your CSV file path
df = pd.read_csv(file_path)
```

```python
keep_list = [
    "Sweater", "T-shirt", "Top", "Blouse", "Shirt", "Vest top", "Hoodie", "Polo shirt",
    "Trousers", "Shorts", "Skirt", "Leggings/Tights", "Outdoor trousers",
    "Jacket", "Cardigan", "Coat",
    "Pyjama set", "Pyjama jumpsuit/playsuit", "Pyjama bottom", "Night gown", "Robe",
    "Blazer", "Outdoor Waistcoat", "Tie", "Tailored Waistcoat",
    "Jumpsuit/Playsuit", "Dungarees", "Outdoor overall", "Dress",
    "Sneakers", "Boots", "Sandals", "Other shoe", "Slippers", "Heeled sandals", "Pumps",
    "Flat shoe", "Flip flop", "Wedge", "Bootie", "Heels", "Flat shoes",
    "Bag", "Weekend/Gym bag", "Backpack", "Cross-body bag", "Tote bag", "Shoulder bag", "Bumbag", "Wallet",
    "Hat/beanie", "Cap/peaked", "Hat/brim", "Beanie", "Cap", "Felt hat", "Straw hat",
    "Earring", "Earrings", "Other accessories", "Necklace", "Ring", "Accessories set", "Bracelet",
    "Watch", "Hair/alice band", "Hair clip", "Hair string", "Hair ties", "Hairband",
    "Sunglasses", "Eyeglasses",
    "Scarf", "Belt", "Gloves", "Socks"
]

df_cleaned = df[df['product_type_name'].isin(keep_list)]

df_cleaned.to_csv('new_cleaned_hm_data.csv', index=False)

print("Filtered data saved to 'new_cleaned_hm_data.csv'.")
```
