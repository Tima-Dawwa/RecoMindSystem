```python
!pip install -U datasets huggingface_hub fsspec

from pathlib import Path
from datasets import load_dataset
from tqdm.auto import tqdm
import os

DATASETS = [
    # "AntZet/men_women_children_wearing_clothes", # not working
    "wbensvage/clothes_desc", # done
    "tomytjandra/h-and-m-fashion-caption", # done
    # "chussboi96/clothes-dataset", # not working
    "hahminlew/kream-product-blip-captions" # done
]

SAVE_DIR = Path("/content/drive/MyDrive/RecoMind/Raw Data")
SAVE_DIR.mkdir(parents=True, exist_ok=True)
```

```python
for ds_id in DATASETS:
    name = ds_id.split("/")[-1]
    out_path = SAVE_DIR / name

    if out_path.exists():
        print(f"{name} already downloaded")
        continue

    print(f"Downloading {ds_id} ...")
    try:
        ds = load_dataset(ds_id, split="train")
        ds.save_to_disk(str(out_path))
        print(f"   -> {len(ds)} rows saved to {out_path}")
    except Exception as e:
        print(f"Failed to download {ds_id}: {e}")
```
