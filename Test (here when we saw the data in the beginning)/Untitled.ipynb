```python
import pandas as pd

# Load the original CSV file
file_path = 'articles.csv'  # Update with your CSV file path
df = pd.read_csv(file_path)

# Select only the required columns
columns_to_keep = ['article_id', 'prod_name', 'product_type_name', 'graphical_appearance_name', 
                   'colour_group_name', 'department_name', 'index_group_name', 'detail_desc']

filtered_df = df[columns_to_keep]

filtered_file_path = 'filtered_data.csv'  # Name of the new CSV file
filtered_df.to_csv(filtered_file_path, index=False)

print(f"Filtered data saved to {filtered_file_path}")
```
