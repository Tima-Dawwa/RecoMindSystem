```python
import os
import pandas as pd

# Load the style IDs from the CSV
csv_path = "C:/fashion-dataset/style_ar.csv"  # change if needed
image_folder = "C:/fashion-dataset/images"     # change if needed

# Read the CSV file
df = pd.read_csv(csv_path)

# Make sure the column name matches the one in your CSV (e.g., 'id')
valid_ids = set(str(id).strip() for id in df['id'])

# Loop through the image files in the folder
for filename in os.listdir(image_folder):
    file_id, ext = os.path.splitext(filename)
    if file_id not in valid_ids:
        # Delete the file
        file_path = os.path.join(image_folder, filename)
        os.remove(file_path)
        print(f"Deleted: {filename}")

print("Cleanup complete.")
```
